**Objetivo**: Visualizar diferentes funciones de activación.

El siguiente código grafica las funciones de activación sigmoide, tangente hiperbólica, ReLU, y leaky ReLU.

```python
import matplotlib.pyplot as plt

import torch
import torch.nn.functional as F
```

```python
def sigmoid(x):     return torch.sigmoid(x)
def tanh(x):        return torch.tanh(x)
def relu(x):        return F.relu(x)
def leaky_relu(x):  return F.leaky_relu(x, negative_slope=0.1)
```

```python
x = torch.linspace(-3, 3, 500)
```

```python
plt.figure(figsize=(11, 7.5))
plt.subplot(2, 2, 1)
plt.plot(x, sigmoid(x), color="#4ab8c8", lw=3)
plt.title("Sigmoid")
plt.subplot(2, 2, 2)
plt.plot(x, tanh(x), color="#6abf5e", lw=3)
plt.title("Hyperbolic Tangent")
plt.subplot(2, 2, 3)
plt.plot(x, relu(x), color="#d94f3a", lw=3)
plt.title("Rectified Linear Unit (ReLU)")
plt.subplot(2, 2, 4)
plt.plot(x, leaky_relu(x), color="#e88080", lw=3)
plt.title("Leaky ReLU")
plt.show()
```
