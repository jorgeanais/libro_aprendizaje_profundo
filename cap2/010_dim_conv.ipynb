**Objetivo**: Esta herramienta te permite calcular el tamaño del mapa de características a partir de las dimensiones de la imágen de entrada, el kernel, el tamaño del padding y stride.



Visualizador interactivo de la fórmula para calcular el tamaño del resultado de una convolución, de acuerdo con la siguiente ecuación:

$$O = \left\lfloor \frac{W - K + 2P}{S} \right\rfloor + 1$$

| Símbolo | Significado |
|---------|------------|
| **W** | Tamaño de entrada |
| **K** | Tamaño del Kernel |
| **P** | Padding |
| **S** | Stride |
| **O** | Tamaño de salida |
